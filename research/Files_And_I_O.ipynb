```python
with open("salam.txt",'rt') as file:
    data = file.read()
for line in data.splitlines():
    print(line)
```

```python
with open("write.txt",'wt') as file:
    file.write("Hello World")
    file.write("\n")
    file.write("Come again!")
    file.write("\n")
    file.close()
```

```python
with open("write.txt",'at') as f:
    print("This is appending to the text file!", file=f)
    print("This is second line to append to the text file!", file=f)
    
```
